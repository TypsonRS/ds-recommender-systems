# Evaluation Metrics, Methods and Concepts

Evaluating a recommender system is not a straightforward task. Its success depends heavily on the people the system serves. It is relatively easy to optimise the underlying machine learning models, but in the end the user makes the decision, and it is hard to say whether a person considers a recommendation good enough. This is why many practitioners regard recommender systems as much an art as a science.

In this notebook you will get an overview of the metrics and methods used to evaluate a recommender system.

## Learning Objectives

At the end of this notebook, you should be able to:

- Explain how error-based accuracy metrics (MAE and RMSE) measure rating-prediction error.
- Compare hit-rate metrics (hit rate, ARHR, cumulative and rating hit rate) for top-N recommendations.
- Distinguish beyond-accuracy metrics (coverage, diversity, novelty, churn) and explain what each captures.
- Select appropriate offline and online methods to evaluate a given recommender.

Evaluation happens at two stages. **Offline metrics** are computed on held-out ratings before the system goes live: they are fast and repeatable, but they only measure what the past data can show. **Online metrics** are measured on real users after release, through A/B tests and direct feedback: they are the ground truth, but they are slower and more expensive to gather.

A common workflow uses offline metrics to shortlist a few models, then confirms the winner online.

## Offline metrics

Offline metrics are computed on held-out ratings, before the system goes live. As in any other machine learning setting, the data is split into a training set and a test set, and the metrics score how well the model reproduces the held-out behaviour.

The metrics below fall into three families:

- **Accuracy metrics** (`MAE`, `RMSE`): how close the predicted ratings are to the true ratings.
- **Hit-rate metrics** (hit rate, `ARHR`, `cHR`, `rHR`): whether the top-N list contains items the user actually wanted.
- **Beyond-accuracy metrics** (coverage, diversity, novelty, churn): qualities of the recommendations that accuracy alone does not capture.

The first two metrics are **accuracy metrics**: they compare a predicted rating with the true rating the user later gave, answering a narrow question, how good is the model at guessing the number?


### MAE

The mean absolute error (`MAE`) sums the absolute difference between each observed rating and its predicted value, then averages over the number of observations. Every error contributes in proportion to its size.

$$
  MAE = \frac{\sum_{i=1}^n |y_{i}-\hat{y_{i}}|}{n}
$$

### RMSE

The root mean squared error (`RMSE`) is the square root of the averaged squared differences between observation and prediction. Because the errors are squared before averaging, `RMSE` penalises large misses more heavily than `MAE` does.

$$
   RMSE = \sqrt{\frac{\sum_{i=1}^n (y_{i}-\hat{y_{i}})^2}{n}}
$$

### Hit rate

The next four metrics are **hit-rate metrics**. They judge the top-N list rather than individual rating predictions: did we recommend something the user actually engaged with?

The user is shown the top-N recommendations. If the user actually rates one of those recommendations, it counts as a hit. The hit rate is the number of hits in the test set divided by the number of users.

Hit rate is usually measured with **leave-one-out cross validation**: for each user, one known item is removed, the top-N recommendations are computed, and we check whether the removed item would have been recommended (a hit). A limitation is gathering enough evaluation data for any single item, so this method suits large data sets.

$$
  hit\ rate = \frac{n_{hits}}{n_{users}}
$$


### Average reciprocal hit rate (ARHR)

The `ARHR` takes the rank of the hit into account: a hit near the top of the list counts for more than a hit further down. This is useful when users have to scroll to find items in the top-N list.

$$
  ARHR = \frac{\sum_{i=1}^{n_{hits}} \frac{1}{rank_i}}{n_{users}}
$$

### Cumulative hit rate (cHR)

The cumulative hit rate (`cHR`) ignores items whose predicted rating falls below a chosen threshold. Only confidently recommended items in the top-N list can count as a hit, so weak predictions do not inflate the score.

### Rating hit rate (rHR)

The rating hit rate (`rHR`) breaks the hit rate down by predicted rating for the items in the top-N list. Because those ratings come from the model, this metric is sensitive to the quality of the predicted ratings, and therefore to the `RMSE` of the model.

### Coverage

The last four metrics are **beyond-accuracy metrics**. Accuracy and hit rate say nothing about the breadth, freshness or variety of what is recommended, and these metrics capture those qualities.

Coverage is the percentage of possible recommendations the system is able to provide: the share of items the system can meaningfully recommend relative to the total number of items. Low coverage means large parts of the catalogue are never surfaced.


### Diversity

Diversity is defined as `1` minus the average similarity between the pairs of recommended items. If diversity is too low, users may get bored and engagement can fall. A more diverse list also helps users discover niche items they would not otherwise find.

### Novelty

Novelty measures how new or unexpected the recommended items are. Users often value recent items, but new items may not have been rated yet, so a system that leans too hard on novelty risks surfacing items it knows little about. Novelty therefore has to be balanced against confidence in the recommendation.

### Churn

Churn measures how much the recommendations for a user change over time. Very low churn means the list looks static and stale, while very high churn means it never settles. A healthy system sits between the two.

## Online metrics

Online metrics are measured on real users after the system goes live. They are the ground truth for whether recommendations work, but they are slower and more expensive to gather than offline metrics.


### A/B tests

An A/B test is the method that matters most when developing recommender systems, because it measures real user behaviour. Two versions of the system (for example, the current recommender and a candidate replacement) are served to comparable groups of users at the same time, and their behaviour is compared to decide which performs better.

### Perceived feedback

Perceived feedback is direct feedback gathered from the user, for example ratings, thumbs up or down, or short surveys about whether a recommendation was helpful. It captures how good the recommendations feel to the user, which the behavioural signals in an A/B test do not always reveal.
