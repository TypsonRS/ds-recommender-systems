## Why a Most-Popular Baseline?

A most-popular recommender is a non-personalised baseline: it ranks items by how often they have been rated (or purchased, or viewed) and recommends the same top items to everyone. It is quick to build, easy to explain, and gives a sensible reference point that any more advanced model should beat.

Its main limitation is that it ignores individual taste. Because every user sees the identical list, the recommendations are not tailored to what a particular person actually likes. This is exactly where personalised recommenders (content-based or collaborative filtering) add value.

The popularity baseline is also a common answer to the cold-start problem: when a new user arrives with no rating history, there is nothing personal to learn from yet, so recommending broadly popular items is a reasonable first response until enough of their interactions have been collected.
